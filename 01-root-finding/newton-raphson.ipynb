#### Newton-Rhapson

```python
import numpy as np
```

```python
def func(x):
    return x**2 - 2
 
# Derivative of the above function
# which is 3*x^x - 2*x
def derivFunc(x):
    return 2*x
 
# Function to find the root
def newtonRaphson(x):
    h = func(x) / derivFunc(x)
    while abs(h) >= 0.0001:
        h = func(x)/derivFunc(x)
         
        # x(i+1) = x(i) - f(x) / f'(x)
        x = x - h
     
    print("The value of the root is : ",
                             "%.10f"% x)
 
# Driver program to test above
x0 = 4.8151 # Initial values assumed
newtonRaphson(x0)
```
