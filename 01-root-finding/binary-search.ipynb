```python
import numpy as np
```

```python
def f(x):
    return x - 4*np.tanh(x)

def binary_search(f, accuracy, x1, x2):
    while True:
        x_line = 0.5*(x1 + x2)
        if np.sign(f(x_line)) == np.sign(f(x1)):
            x1 = x_line
        else:
            x2 = x_line
        if abs(x1-x2) < accuracy:
            x_line = 0.5*(x1+x2)
            return x_line
        
x_line = binary_search(f, 0.0076171, 0.1, 4)
print(x_line)
```
