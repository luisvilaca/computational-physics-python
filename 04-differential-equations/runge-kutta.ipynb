```python
def function(r, t, alpha, beta, gamma, delta):
    """r[0] = x, r[1] = y"""
    return np.array([alpha * r[0] - beta * r[0]*r[1], gamma*r[0]*r[1] - delta*r[1]], float)

def RungeKutta(F, T, N, R_0, alpha, beta, gamma, delta):
    tempos = np.linspace(0, T, N)
    r = np.zeros([N,2], float)
    r[0,:] = R_0
    h = tempos[1] - tempos[0]
    for i in range(N-1):
        k1 = h * F(r[i,:], tempos[i], alpha, beta, gamma, delta)
        k2 = h * F(r[i,:] + 0.5*k1, tempos[i] + 0.5*h, alpha, beta, gamma, delta)
        k3 = h * F(r[i,:] + 0.5*k2, tempos[i] + 0.5*h, alpha, beta, gamma, delta)
        k4 = h * F(r[i,:] + k3, tempos[i] + h, alpha, beta, gamma, delta)
        r[i+1,:] = r[i,:] + (k1+2*k2+2*k3+k4)/6
    return tempos, r
```
