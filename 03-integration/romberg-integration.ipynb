```python
def f(x):
    return x**6 + x**5 + 3*x**4 + 4*x**2 + 3*x + 5

def Trapezio(f, a, b, N):
    """Método Integração Trapézio"""
    sum = 0.5*f(a) + 0.5*f(b)
    h = (b - a) / N
    for k in range(1, N):
        sum += f(a + k*h)
    return (sum * h)
    
def Romberg(f, a, b, p, epsilon, N):
    """Método Integração Romberg, R(i,m) com ordem de erro = 2m+2. N = nº de amostras, a e b são os limites de integração. P é o limite de linhas
    e colunas do método de Romberg."""
    I = np.zeros((p,p), float)
    for i in range (1,len(I+1)):
        for m in range (1,i+1):
            if i == 1 and m == 1:
                # Não é possível calcular o erro, cálculo do primeiro integral
                I[i,m] = Trapezio(f, a, b, N)
            elif i != 1 and m == 1:
                # Cálculo do Integral para novo passo no método de Romberg, dobro do Nº de Amostras
                I[i,m] = Trapezio(f, a, b, 2**(i-1) * N)
                error = (1 / (4**m - 1)) * (I[i,m] - I[i-1,m])
                if abs(error) <= epsilon:
                    return I[i,m]
            elif m != 1 and i != m:
                I[i,m] = I[i,m-1] + (1 / (4**(m-1) - 1)) * (I[i,m-1] - I[i-1,m-1])
                error = (1 / (4**m - 1)) * (I[i,m] - I[i-1,m])
                if abs(error) <= epsilon:
                    return I[i,m], error
            elif m != 1 and m == i:
                #Impossível calcular o erro para o último elemento de cada linha
                I[i,m] = I[i,m-1] + (1 / (4**(m-1) - 1)) * (I[i,m-1] - I[i-1,m-1])
                                                            
print(Romberg(f,0,2,7,0.000001,2))
```
