#### Integral de 0 -> $\infty$

Tendo o seguinte integral, e fazendo a mudança de varíavel z:

$$\int\limits_{0}^{\infty}f(x)dx$$

$$z = \frac{x}{1+x}$$

$$x = \frac{z}{1-z}$$

$$dx = \frac{dz}{(1-z)^2}$$

De acordo com esta mudança de variável, o integral fica:

$$\int\limits_{0}^{\infty}f(x)dx = \int\limits_0^1 \frac{1}{(1-z)^2} f \bigg(\frac{z}{(1-z)}\bigg) dz$$

#### Integral de a -> $\infty$

$$y = x-a$$

$$z = \frac{y}{1+y} = \frac{x-a}{1+x-a}$$

Logo, assim temos:

$$x = \frac{z}{1-z} + a$$

$$dx = \frac{dz}{(1-z)^2}$$

Concluindo, obtemos:

$$\int\limits_{a}^{\infty}f(x)dx = \int\limits_{0}^{1} \frac{1}{(1-z)^2} f\Big(\frac{z}{1-z}+a\Big)dz$$

#### Integral de $-\infty$ -> a

Realizar o mesmo procedimento acima, substituíndo z -> -z

#### Integral de $-\infty$ -> $\infty$

Realizando a seguinte mudança de variável

$$x = \frac{z}{1-z^2}$$

$$dx = \frac{1+z^2}{\left(1-z^2\right)^2}dz$$

Obtem-se:

$$\int\limits_{-\infty}^{\infty}dxf(x) = \int\limits_{-1}^{1}dz \frac{1+z^2}{\left(1-z^2\right)^2} f\left(\frac{z}{1-z^2}\right)$$

Ou então, realizando uma mudança de variável mais simples:

$$x = tan(z)$$

$$dx = \frac{dz}{cos^2(z)}$$

Logo, o integral toma a seguinte forma:

$$\int\limits_{-\infty}^{\infty} f(x)dx = \int\limits_{-\pi/2}^{\pi/2} \frac{1}{cos^2(z)} f(tan(z)) dz$$

**NOTA:**
Deve-se recorrer ao método da quadratura gaussiana para conseguir realizar os integrais impróprios, pois existe erro de divisão por zero utilizando os restantes métodos.
