#### Método Trapézio

```python
def Trapezio(f, a, b, N):
    """Método Integração Trapézio"""
    sum = 0.5*f(a) + 0.5*f(b)
    h = (b - a) / N
    for k in range(1, N):
        sum += f(a + k*h)
    return (sum * h)
```

#### Método Trapézio Adaptativo

```python
def Trapezio_Adap(f, a, b, epsilon):
    """Método Integração Trapézio Adaptativo"""
    N = 1 # Nº amostras estimativa inicial
    I_1 = Trapezio(f, a, b, N) # Estimativa inicial
    i = 1
    while True:
        I_2 = Trapezio(f, a, b, 2**i * N)
        error = (1/3) * (I_2 - I_1)
        if abs(error) <= epsilon:
            return I_2
        else:
            I_1 = I_2
            i += 1
```
