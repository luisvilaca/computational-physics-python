```python
import numpy as np
```

```python
def f(x):
    return 5*x

def Simpson(f, a, b, N):
    """Método de Simpson para Integração"""
    h = (b - a) / N
    sum = 0
    #Soma de termos limite
    sum += f(a) + f(b)
    #Soma de termos ímpares
    for k in range(1, N, 2):
        sum += 4 * f(a + k*h)
    #Soma de termos pares
    for j in range(2, N-1, 2):
        sum += 2 * f(a + j*h)
    sum *= 1/3 * h
    return sum

print(Simpson(f, 0, 5, 10))

def Simpson_w(f, a, b, N):
    """Returns weights and points"""
    h = (b - a) / N
    w = np.ones(N+1)
    x = np.linspace(a, b, N+1)
    for k in range(1, N+1, 2):
        w[k] = 4
    for j in range(2, N, 2):
        w[j] = 2
    w *= (1/3) * h
    return x,w

N = 2
a = 0
b = 5
h = (b-a) / N
x, w = Simpson_w(f, a, b, N)
I = 0
for k in range(len(x)):
    I += w[k]*f(x[k])
print(I)
```
