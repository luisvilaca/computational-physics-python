#### Integrais Múltiplos 

Considerando o seguinte integral múltiplo:

$$I = \int\limits_{0}^{1}\int\limits_{0}^{1}f(x,y)dxdy$$

Considerando uma função F(y) tal que:

$$F(y) = \int\limits_{0}^{1}f(x,y)dx$$

Logo, o integral será igual:

$$I = \int\limits_{0}^{1}F(y)dy$$

Pelo método da quadratura de Gauss é possível calcular o integral da seguinte forma, escolhendo o mesmo número de amostras para cada somatório:

$$F(y) = \sum\limits_{i=1}^{N}w_i f(x_i,y)$$
$$I = \sum\limits_{j=1}^{N}w_j F(y_j)$$

Logo, pode-se sumarizar desta forma:

$$I = \sum\limits_{i=1}^{N} \sum\limits_{j=1}^{N}w_i w_j f(x_i,y_j)$$

#### Exemplo

Resolver o seguinte integral:

$$I = \int\limits_{0}^{5}\int\limits_{2}^{4}(xy^3 - yx^2)dxdy$$

```python
import numpy as np
```

```python
def gaussxw(N):
    # Gauss-Legendre sample points and weights on [-1, 1]
    return np.polynomial.legendre.leggauss(N)

def gaussxwab(N, a, b):
    # Same points and weights mapped to [a, b]
    x, w = gaussxw(N)
    return 0.5*(b-a)*x + 0.5*(b+a), 0.5*(b-a)*w
```

```python
def f(x,y):
    return x * y**2 - y * x

sum = 0
N = 20
xi, wi = gaussxwab(N,0,10)
yj, wj = gaussxwab(N,0,5)

for i in range(N):
    for j in range(N):
        sum += wi[i] * wj[j] * f(xi[i],yj[j])
        
print(sum)
```

```python
def F(x,y):
    return x * y**2 - y * x

def IntegrationX(Nx, xi, xf, y):
    x = np.linspace(xi, xf, Nx)
    h_x = x[1] - x[0]
    return h_x/3 * (F(x[0], y) + F(x[-1], y) + 4*np.sum(F(x[1:-1:2], y)) + 2*np.sum(F(x[2:-2:2], y))) 
    
def Integration(Nx, Ny, xi, xf, yi, yf):
    y = np.linspace(yi, yf, Ny)
    h_y = y[1] - y[0]
    func = np.vectorize(lambda ys: IntegrationX(Nx, xi, xf, ys))
    f = func(y)
    return h_y/3 * (f[0] + f[-1] + 4*np.sum(f[1:-1:2]) + 2*np.sum(f[2:-2:2])) 
    
print(Integration(10000, 10000, 0, 10, 0, 5))
```
