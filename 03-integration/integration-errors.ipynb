#### Erro Integração Trapézio

$\epsilon = \frac{h^2}{12} * [f'(a) - f'(b)]$

##### Nº de Passos Máximo

$\frac{h^2}{12} * [f'(a) - f'(b)] = C*\int\limits_a^b{f(x)}dx $

$h = \sqrt{\frac{12*\int\limits_a^b{f(x)}dx}{f'(a) - f'(b)}} * C^\frac{1}{2}$

$N = (b-a)*\sqrt{\frac{f'(a) - f'(b)}{12*\int\limits_a^b{f(x)}dx}} * C^-\frac{1}{2}$ 

#### Erro Integração Simpson

$\epsilon = \frac{h^4}{90} * [f'''(a) - f'''(b)]$

##### Nº de Passos Máximo

$\frac{h^4}{90} * [f'''(a) - f'''(b)] = C*\int\limits_a^b{f(x)}dx$

$N = (b - a) * \sqrt[4]{\frac{f'''(a)-f'''(b)}{90*\int\limits_a^b{f(x)dx}}} * C^-\frac{1}{4}$

#### Cálculo Prático do Erro com Amostras

##### Trapézio

$\epsilon_2 = \frac{1}{3} * (I_2-I_1)$, $h_1 = 2*h_2$

##### Simpson

$\epsilon_2 = \frac{1}{15} * (I_2 - I_1), h_1 = 2*h_2$

#### Erro Integração Gauss

No método de Gauss não é possível calcular o erro através de uma fórmula como no método de Simpson e do Trapézio, contudo sabe-se que o erro está relacionado com o número de amostras N através da seguinte fórmula:

$\epsilon_{N} = \frac{c}{N^2}$, sendo c um número arbitrário dependente da forma do integral e do seu domínio.

Considerando:

$I = I_1 + \epsilon_{N},$
$I = I_2 + \epsilon_{2N}$

Para um número N suficientemente grande:

$\epsilon_N >> \epsilon_{2N}$

$\epsilon_N = I_2 - I_1$
