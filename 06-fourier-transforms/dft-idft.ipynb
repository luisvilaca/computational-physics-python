```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def DFT(y):
    N = len(y)
    n = np.arange(N)
    c = np.zeros(N, complex)
    for k in range(N):
        c[k] = np.sum(y*np.exp(-2j*np.pi*k*n / N))
    return c
    
y = [2,4,6,1,3,5,6]
c = DFT(y)
print(c)
plt.plot(y,c)
```

```python
def IDFT(c):
    N = len(c)
    k = np.arange(N)
    y = np.zeros(N, complex)
    for n in range(N):
        y[n] = 1/N * np.sum(c*np.exp(2j*np.pi*k*n / N))
    return y

c = DFT(y)
print(c)
y_1 = IDFT(c)
print(y_1)
print(abs(y_1))
```
