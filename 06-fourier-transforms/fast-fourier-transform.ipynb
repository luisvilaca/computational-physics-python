```python
x = sympy.Symbol('x')
y = sympy.Function('y')
piI = sympy.pi * sympy.I



M = sympy.Matrix([[y(int('{:03b}'.format(i)[::-1], 2))] for i in range(8)])

display(M)

for i in range(4):
    e = M[2*i, 0]
    o = M[2*i + 1, 0]
    M[2*i, 0] =     e + sympy.exp(-piI * 2/2 * 0 ) * o
    M[2*i + 1, 0] = e -  sympy.exp(-piI * 2/2 * 0 ) * o

M = M.reshape(4,2)
display(M)

for k in range(2):
    for i in range(2):
        e = M[2*i, k  ]
        o = M[2*i + 1, k]
        M[2*i, k] = e + sympy.exp(-piI * 2 /4*k  ) * o
        M[2*i + 1, k] = e -  sympy.exp(-piI * 2/4 * k  ) * o
M = M.reshape(2,4)
display(M)

for j in range(4):
    for i in range(1):
        e = M[2*i, j]
        o = M[2*i + 1, j]
        M[2*i, j] = e + sympy.exp(-piI * 2/8 * j ) * o
        M[2*i + 1, j] = e +  sympy.exp(-piI * 2 /8 * (j + 4) ) * o
M = M.reshape(1,8)
display(M)
```

```python
x = sympy.Symbol('x')
i1 = sympy.Wild('i1')
for k in range(8):
    i = sympy.Symbol('i')
    exact_sum = sympy.Sum( y(i) * sympy.exp(-2 * piI / 8 * i * k ),(i,0,7))
    
    aa = M[0, k].expand() - exact_sum
    aa = aa.simplify().trigsimp().cancel()
    display(aa.replace( (-1)**i1 , sympy.exp(sympy.pi * sympy.I * i1)))  
```

```python
expoente = 4
L = 2**expoente
x = sympy.Symbol('x')
y = sympy.Function('y')

pi2I = 2*sympy.pi * sympy.I
name = '{:0'+str(expoente)+'b}'
M = sympy.Matrix([[y(int(name.format(i)[::-1], 2))] for i in range(L)])

display(M)
for est in range(expoente):
    ll = 2**(est + 1)
    for k in range(ll//2):
        for i in range(L//ll):
            e = M[2*i, k]
            o = M[2*i + 1, k]
            M[2*i, k] = e + sympy.exp(-pi2I/ll * k ) * o
            M[2*i + 1, k] = e -  sympy.exp(-pi2I/ll * k  ) * o
    M = M.reshape(L//ll, ll)
    display(M)


x = sympy.Symbol('x')
i1 = sympy.Wild('i1')
for k in range(L):
    i = sympy.Symbol('i')
    exact_sum = sympy.Sum(y(i) * sympy.exp(-pi2I / L * i * k ), (i,0,L-1))
    aa = M[0, k].expand() - exact_sum
    aa = aa.replace(y(i1), lambda i1: x**i1).simplify().trigsimp().cancel()
    display(aa.replace( (-1)**i1 , sympy.exp(sympy.pi * sympy.I * i1)))  
```

```python
def myFFT(y):
    M = len(y)
    m = int(np.round(np.log2(M)))
    c = np.zeros((M, 1), complex)
    name = '{:0'+str(m)+'b}'
    
    
    for i in range(M):
        i1 = int(name.format(i)[::-1], 2)
        c[i,0] = y[i1]

        
    for m1 in range(m):
        ll = 2**(m1 + 1)
        for k in range(ll//2):
            f = np.exp(- 2j*np.pi/ll * k )
            for n in range(M//ll):
                par   = c[2 * n, k]
                impar = c[2 * n + 1, k]
                c[2 * n, k], c[2 * n + 1, k] = par + f * impar, par - f * impar
        c = c.reshape(M//ll , ll)
    return c
         

def DFT(y):
    M = len(y)
    c = np.zeros((M, 1), complex)
    
    for k in range(M):
        for n in range(M):
            c[k] += np.exp(-2j*np.pi*k * n/M)*y[n]
    return c

    
    
k1 = 4
L1 = 2**8
y1 = np.exp(2j*np.pi*np.arange(L1) * k1/L1 )

c1 = DFT(y1)
plt.plot(np.real(c1).flatten())
```
