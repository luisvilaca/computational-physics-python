#### Decomposição QR

```python
def QR_decomp(A):
    """QR Decomp, retorna Q como primeiro argumento e R como segundo"""
    N = len(A)
    u = np.zeros((N,N), float)
    Q = np.zeros((N,N), float)
    for i in range(N):
        u[:,i] = A[:,i]
        for k in range(i):
            u[:,i] -= np.dot(Q[:,k], A[:,i]) * Q[:,k]
        Q[:,i] = u[:,i] / np.linalg.norm(u[:,i])
    R = np.zeros((N,N), float)
    for l in range(N):
        R[l,l] = np.linalg.norm(u[:,l])
        for m in range(l+1, N):
            R[l,m] = np.dot(Q[:,l], A[:,m])
    return Q,R

A = np.array([[ 0,  3,  1,  0,  8, 6],
           [ 3,  4, 8, 8, 0 , 3],
           [ 8, 2,  2,  5, 5, 1 ],
           [ 0, 3,  9,  1, 7, 2 ],
            [8, 2, 3, 0, 2, 1],
             [4, 7, 1, 0, 5 ,6]],float)
Q,R = QR_decomp(A)
print(R)
```

```python
def QR_householder(A):
    """QR Decomp, utilizando o método de householder, retorna matrizes Q e R"""
    (num_rows, num_cols) = np.shape(A)
    Q = np.identity(num_rows)
    R = np.copy(A)
    for cnt in range(num_rows - 1):
        x = R[cnt:, cnt]
        e = np.zeros_like(x)
        e[0] = np.copysign(np.linalg.norm(x), -A[cnt, cnt])
        u = x + e
        v = u / np.linalg.norm(u)
        Q_cnt = np.identity(num_rows)
        Q_cnt[cnt:, cnt:] -= 2.0 * np.outer(v, v)
        R = np.dot(Q_cnt, R)
        Q = np.dot(Q, Q_cnt.T)
    return (Q, R)

def eigen(A, error):
    """Retorna valores próprios e vetores próprios"""
    N = len(A)
    V = np.identity(N, float)
    while True:
        Q, R = QR_householder(A)
        A = np.dot(R,Q)
        V = np.dot(V,Q)
        # Verificar valores fora das diagonais
        max_dev = 0
        for i in range(N):
            for j in range(N):
                if i == j:
                    continue
                else:
                    if A[i,j] >= max_dev:
                        max_dev = A[i,j]
        # Comparar desvio máximo com o erro
        if max_dev <= error:
            eigenvalues = np.zeros(N, float)
            for i in range(N):
                eigenvalues[i] = A[i,i]
            return eigenvalues,V
            break

#Definir matriz A
N = 6
A = np.zeros((N,N), float)
A[0][0] = 2
A[0][1] = -1
A[N-1][N-2] = -1
A[N-1][N-1] = 1
for i in range(1,N-1):
    A[i][i-1] = -1
    A[i][i] = 2
    A[i][i+1] = -1

print(A)
eig, v = eigen(A, 10**-5)
print(eig)
print(v)
print(v[0,:])
v[0,:] = v[0,:] / np.linalg.norm(v[0,:])
print(v[0,:])
newvec = v[0,:] * np.conjugate(v[0,:])
sum = np.sum(newvec)
print(sum)
```
