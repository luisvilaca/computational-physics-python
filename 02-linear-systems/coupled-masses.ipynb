```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def elim_gauss(A, v):
    N = len(v)
    # Gaussian Elimination
    for i in range(N):
        div = A[i,i]
        A[i, :] /= div
        v[i] /= div
        for j in range(i+1, N):
            mult = A[j,i]
            A[j,:] -= mult * A[i,:]
            v[j] -= mult * v[i]
    # Backsubstituion
    x = np.empty(N, float)
    for k in range(N-1, -1, -1):
        x[k] = v[k]
        for i in range(k+1, N):
            x[k] -= A[k,i] * x[i]
    return x

#Constantes
N = 7
C = 5
k = 6.0
m = 1.0
omega = 2.0
alpha = 2*k - m*omega**2

#Construção da matriz A
A = np.zeros((N,N), float)
for i in range(1, N-1):
    A[i,i-1] = -k
    A[i,i] = alpha
    A[i,i+1] = -k
A[0,0] = alpha - k
A[0,1] -= k
A[N-1,N-2] -= k
A[N-1,N-1] = alpha - k
#Construção do vetor V
v = np.zeros(N, float)
v[0] = C
print(A)

x = elim_gauss(A,v)
print(x)
plt.plot(x)
plt.plot(x, "ko")
plt.xlabel("Nº massa")
plt.ylabel("Posição")
```
