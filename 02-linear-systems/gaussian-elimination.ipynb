```python
import numpy as np
```

#### Eliminação Gaussiana

```python
def elim_gauss(A, v):
    N = len(v)
    # Gaussian Elimination
    for i in range(N):
        div = A[i,i]
        A[i, :] /= div
        v[i] /= div
        for j in range(i+1, N):
            mult = A[j,i]
            A[j,:] -= mult * A[i,:]
            v[j] -= mult * v[i]
    # Backsubstituion
    x = np.empty(N, float)
    for k in range(N-1, -1, -1):
        x[k] = v[k]
        for i in range(k+1, N):
            x[k] -= A[k,i] * x[i]
    return x

A = np.array([[ 3,  1,  4,  1 ],
           [ 3,  4, -1, -1 ],
           [ 1, -4,  1,  5 ],
           [ 2, -2,  1,  3 ]],float)
v = np.array([ -4, 3, 9, 7 ],float)
x = elim_gauss(A, v)
print(x)
```

#### Eliminação Gaussiana com Pivotagem

```python
def elim_gauss(A, v):
    N = len(v)
    # Gaussian Elimination
    for i in range(N):
        # Pivotagem Parcial
        for m in range(i+1, N):
            if abs(A[m,i]) > abs(A[i,i]):
                # Troca linhas na matriz
                temp = np.copy(A[i,:])
                A[i,:] = A[m,:]
                A[m,:] = temp
                # Troca de elementos no vetor
                temp1 = np.copy(v[i])
                v[i] = v[m]
                v[m] = temp1
        # Normalização do pivot e subtração nas linhas
        div = A[i,i]
        A[i, :] /= div
        v[i] /= div
        for j in range(i+1, N):
            mult = A[j,i]
            A[j,:] -= mult * A[i,:]
            v[j] -= mult * v[i]
    # Backsubstituion
    x = np.empty(N, float)
    for k in range(N-1, -1, -1):
        x[k] = v[k]
        for i in range(k+1, N):
            x[k] -= A[k,i] * x[i]
    return x

A = np.array([[ 0,  1,  4,  1 ],
           [ 3,  4, -1, -1 ],
           [ 1, -4,  1,  5 ],
           [ 2, -2,  1,  3 ]],float)
v = np.array([ -4, 3, 9, 7 ],float)
x = elim_gauss(A, v)
print(x)
```
