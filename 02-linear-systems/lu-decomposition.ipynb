```python
import numpy as np
```

```python
def elim_gauss(A, v):
    N = len(v)
    # Gaussian Elimination
    for i in range(N):
        # Pivotagem Parcial
        for m in range(i+1, N):
            if abs(A[m,i]) > abs(A[i,i]):
                # Troca linhas na matriz
                temp = np.copy(A[i,:])
                A[i,:] = A[m,:]
                A[m,:] = temp
                # Troca de elementos no vetor
                temp1 = np.copy(v[i])
                v[i] = v[m]
                v[m] = temp1
        # Normalização do pivot e subtração nas linhas
        div = A[i,i]
        A[i, :] /= div
        v[i] /= div
        for j in range(i+1, N):
            mult = A[j,i]
            A[j,:] -= mult * A[i,:]
            v[j] -= mult * v[i]
    # Backsubstituion
    x = np.empty(N, float)
    for k in range(N-1, -1, -1):
        x[k] = v[k]
        for i in range(k+1, N):
            x[k] -= A[k,i] * x[i]
    return x

A = np.array([[ 0,  1,  4,  1 ],
              [ 1,  4, -1, -1 ],
              [ 4, -1,  1,  5 ],
              [ 2, -2,  1,  3 ]],float)
v = np.array([2,6,9,4], float)
x = elim_gauss(A, v)
print(x)
```

```python
def LU_decomp(A):
    """Método decomposição LU com pivotagem parcial. Retorna L, U e uma matriz swaps que identifica as trocas necessárias para obter a matriz A original."""
    N = len(A)
    L = np.zeros((N,N), float)
    U = np.zeros((N,N), float)
    swaps = np.identity(N, float)
    # Gaussian Elimination
    for i in range(N):
        # Pivotagem Parcial
        L[i:,i] = np.copy(A[i:,i])
        for m in range(i+1, N):
            if abs(A[m,i]) > abs(A[i,i]):
                # Troca linhas na matriz
                temp = np.copy(A[i,:])
                A[i,:] = A[m,:]
                A[m,:] = temp
                temp1 = np.copy(L[i,:])
                L[i,:] = L[m,:]
                L[m,:] = temp1 
                temp2 = np.copy(swaps[i,:])
                swaps[i,:] = swaps[m,:] 
                swaps[m,:] = temp2
        # Normalização do pivot e subtração nas linhas
        div = A[i,i]
        A[i, :] /= div
        for j in range(i+1, N):
            mult = np.copy(A[j,i])
            A[j,:] -= mult * A[i,:]    
    U = np.copy(A)
    return L, U, swaps

A = np.array([[ 0,  1,  4,  1 ],
              [ 1,  4, -1, -1 ],
              [ 4, -1,  1,  5 ],
              [ 2, -2,  1,  3 ]],float)

L, U, swaps = LU_decomp(A) 
print(L)
print(U)
res = np.dot(L,U)
print(np.dot(swaps,res))
```

```python
def sol_LU(L, U, v):
    """Solução Ax = v, com as matrizes L e U"""
    # Resolução para o vetor y
    N = len(v)
    y = np.empty(N, float)
    for k in range(N):
        y[k] = v[k]
        for i in range(k):
            y[k] -= L[k,i] * y[i]
        y[k] /= L[k,k]
    # Resolução para o vetor x
    x = np.empty(N, float)
    for k in range(N-1, -1, -1):
        x[k] = y[k]
        for i in range(k+1, N):
            x[k] -= U[k,i] * x[i]
    return x


v1 = np.array([2,6,9,4], float)
v2 = np.array([3,8,7,5], float)
v1_swap = np.dot(swaps, v1)
x1 = sol_LU(L, U, v1)
x2 = sol_LU(L, U, v2)
x3 = sol_LU(L, U, v1_swap)
print(x1, x2, x3)
```
