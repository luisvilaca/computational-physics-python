```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def GaussSeidel_Laplace():
    # Constantes
    M = 100    
    V = 1
    target = 1e-6
    dif = 1
    w = 0.9
    
    #Matriz Phi
    phi = np.zeros([M+1,2*M+1])
    phi[:,0] = -V
    phi[:,2*M] = -V
    phi[M,:] = V
    phi[0,:] = -V
    for i in range(1,M):
            for j in range(1,2*M): 
                if j<M-i or j>M+i:
                    phi[i,j] = -V

    while dif > target:
        dif = 0
        for i in range(1,M):
            for j in range(M-i,M+i+1): 
                diferenca = (phi[i+1,j] + phi[i-1,j] + phi[i,j+1] + phi[i,j-1])/4 - phi[i,j]
                phi[i,j] = phi[i,j] + (1+w)*diferenca
                if diferenca > dif :
                    dif = diferenca
    
    plt.figure(figsize = (16,8))
    extent = 0, 10, 0, 5
    plt.imshow(phi, extent = extent)
    plt.colorbar()
    
GaussSeidel_Laplace()
```
