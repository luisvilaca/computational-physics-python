```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def Poisson():
    
    # Constantes
    M = 100
    V = 1.0
    target = 1e-6
    dif = 1
    
    # Arrays necessários
    phi = np.zeros([M+1, M+1])
    rho = np.copy(phi)
    rho[20:40,60:80] = 1
    rho[60:80,20:40] = -1
    phil = np.copy(phi)
    
    while dif > target:
        
        for i in range(1,M):
            for j in range(1,M):
                phil[i,j] = (phi[i+1,j] + phi[i-1,j] + phi[i,j+1] + phi[i,j-1])/4 + 1/4*rho[i,j]
                
        dif = np.max(abs(phi-phil))

        phi,phil = phil,phi
    
    plt.imshow(phi)
    plt.gray()
    plt.colorbar()
    plt.show()

Poisson()
```

```python
def GaussSeidel_Poisson():
    
    # Constantes
    M = 100
    V = 1.0
    target = 1e-6
    dif = 1
    w = 0.9
    
    # Arrays necessários
    phi = np.zeros([M+1, M+1])
    rho = np.copy(phi)
    rho[20:40,60:80] = 1
    rho[60:80,20:40] = -1
    
    while dif > target:
        dif = 0
        
        for i in range(1,M):
            for j in range(1,M):
                
                diferenca = ( (phi[i+1,j] + phi[i-1,j] + phi[i,j+1] + phi[i,j-1])/4 + 1/4*rho[i,j] ) - phi[i,j]
                phi[i,j] = phi[i,j] + (1+w)*diferenca
                
                if diferenca > dif :
                    dif = diferenca
            
    plt.imshow(phi, origin = "lower")
    plt.jet()
    plt.show()

GaussSeidel_Poisson()
```
